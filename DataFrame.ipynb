 Criar um DataFrame com Dados Fictícios

Vamos criar um DataFrame com alguns dados fictícios. Um DataFrame é uma estrutura de dados semelhante a uma tabela, onde podemos armazenar e manipular dados.


```python
import pandas as pd
import matplotlib.pyplot as plt

# Criando um DataFrame com dados fictícios
def criar_dataframe():
    data= {
       'Nome':['Gustavo','Reginaldo','Murilo','Ofélia'],
       'Idade':[24,12,29,36],
       'Salario': [2300,7800,6500,1150]
       }
    return pd.DataFrame(data)

def mostrar_estatisticas(df):
# Exibindo as primeiras linhas do DataFrame
    print("\n Funcionários ordenados pelo salário")
    print(df.sort_values(by='Salario', ascending=False))

    print(f"Média de idade: {df['Idade'].mean():.2f}")
    print(f"Média de Salário: {df['Salario'].mean():.2f}")
    print(f"Soma de Idade: {df['Idade'].sum()}")
    print(f"Soma de Salário: {df['Salario'].sum():.2f}")

def gerar_grafico(df):

    plt.figure(figsize=(8,5))

    barras = plt.bar(df['Nome'], df['Salario'])
    plt.title('Salário dos Funcionários')
    plt.xlabel('Funcionários')
    plt.ylabel('Salário')
    plt.grid(axis='y')


    for barra in barras :
       altura = barra.get_height()

       plt.text(
        barra.get_x() + barra.get_width()/2,
        altura,
        f'R$ {altura}',
        ha='center'
    )

    plt.show()

while True:

    print("\n====== MENU ======")
    print("1 - Mostrar funcionários")
    print("2 - Mostrar médias")
    print("3 - Gerar gráfico")
    print("4 - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == '1':
        print("\nFuncionários:")
        print(df)

    elif opcao == '2':
        mostrar_estatisticas(df)

    elif opcao == '3':
        gerar_grafico(df)

    elif opcao == '4':
        print("Encerrando sistema...")
        break

    else:
        print("Opção inválida.")

df = criar_dataframe()





```

### Calcular a Média e a Soma das Colunas do DataFrame

Agora vamos calcular a média e a soma das colunas 'Idade' e 'Salário'.

```python
media_idade= df['Idade'].mean()
media_salario= df['Salário'].mean()
soma_idade= df['Idade'].sum()
soma_salario= df['Salário'].sum()
print(f"Média de idade: {media_idade}")
print(f"Média de Salário: {media_salario}")
print(f"Soma de Idade: {soma_idade}")
print(f"Soma de Salário: {soma_salario}")
```
